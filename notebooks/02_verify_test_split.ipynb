```python
import json
from pathlib import Path

# Resolve paths dynamically from the notebooks directory
PROJECT_ROOT = Path().resolve().parents[0] 
RAW_DATA_DIR = PROJECT_ROOT / "data" / "raw"
PROCESSED_DATA_DIR = PROJECT_ROOT / "data" / "processed"

# Exact paths to the files we want to compare
original_test_txt = RAW_DATA_DIR / "UCF_Crimes-Train-Test-Split" / "Anomaly_Detection_splits" / "Anomaly_Test.txt"
processed_test_json = PROCESSED_DATA_DIR / "test_split.json"

print(f"Checking original TXT at: {original_test_txt.exists()}")
print(f"Checking processed JSON at: {processed_test_json.exists()}")
```

```python
original_stems = set()

with open(original_test_txt, 'r', encoding='utf-8') as f:
    for line in f:
        parts = line.strip().split()
        if parts:
            # Extract just the base filename (e.g., 'Abuse028_x264')
            original_stems.add(Path(parts[0]).stem)

print(f"Total videos in original Anomaly_Test.txt: {len(original_stems)}")
```

```python
processed_stems = set()

with open(processed_test_json, 'r', encoding='utf-8') as f:
    test_data = json.load(f)

for item in test_data:
    # Extract the stem from the video_path key
    video_path = item.get("video_path")
    if video_path:
        processed_stems.add(Path(video_path).stem)

print(f"Total videos in processed test_split.json: {len(processed_stems)}")
```

```python
missing_in_json = original_stems - processed_stems
extra_in_json = processed_stems - original_stems

print("--- INTEGRITY CHECK RESULTS ---")
if not missing_in_json and not extra_in_json:
    print("✅ PERFECT MATCH!")
    print("The processed test_split.json contains the exact same videos as the original baseline.")
    print("Your test set is 100% SOTA-compliant.")
else:
    print("❌ MISMATCH DETECTED!")
    
    if missing_in_json:
        print(f"\nVideos in Original but missing in JSON ({len(missing_in_json)}):")
        for v in sorted(missing_in_json)[:10]:
            print(f"  - {v}")
        if len(missing_in_json) > 10: print("  - ...")
            
    if extra_in_json:
        print(f"\nVideos in JSON but not in Original ({len(extra_in_json)}):")
        for v in sorted(extra_in_json)[:10]:
            print(f"  - {v}")
        if len(extra_in_json) > 10: print("  - ...")
```
